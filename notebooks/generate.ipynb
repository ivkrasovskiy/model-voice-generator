# IndexTTS-2 — generate Cumberbatch voice from text

**Kernel**: `vendor/index-tts/.venv/bin/python` (select via *Select Kernel* in the top-right).

**Reference**: `tts_output/ref_interview.wav` (14 s interview clip).

Workflow:
1. Run cell 1 once per session — loads the model (~30 s).
2. Edit `TEXT` in cell 2 and re-run to generate any phrase. Audio plays inline.

```python
# Cell 1 — one-time setup. Run once per kernel session.
import os, sys
from pathlib import Path

os.environ.setdefault("PYTHONHASHSEED", "0")
PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
INDEXTTS_ROOT = PROJECT_ROOT / "vendor" / "index-tts"
sys.path.insert(0, str(INDEXTTS_ROOT))

from indextts.infer_v2 import IndexTTS2

REF_AUDIO = str(PROJECT_ROOT / "tts_output/ref_interview.wav")
CFG_PATH  = str(INDEXTTS_ROOT / "checkpoints/config.yaml")
MODEL_DIR = str(INDEXTTS_ROOT / "checkpoints")

print("Loading IndexTTS-2 on CPU (~30 s)...")
tts = IndexTTS2(cfg_path=CFG_PATH, model_dir=MODEL_DIR, device="cpu")
print(f"  Loaded. Reference: {Path(REF_AUDIO).name}")
```

```python
# Cell 2 — edit TEXT and re-run to generate.
import tempfile
from IPython.display import Audio, display

TEXT = "There is nothing more deceptive than an obvious fact."

out_path = Path(tempfile.mkstemp(prefix="indextts_nb_", suffix=".wav")[1])
tts.infer(spk_audio_prompt=REF_AUDIO, text=TEXT, output_path=str(out_path))
print(f"→ {out_path}")
display(Audio(str(out_path), autoplay=False))
```
